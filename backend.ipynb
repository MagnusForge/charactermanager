```python
# Coming from UI
test_payload = {'name':'Magnus',
       'species':'Elf',
       'background':'Crime'}

# Bad payload to test validate_payload
test_faulty_payload = {'bad_name':'Magnus',
       'bad_value': 1234,
       'background':'Crime'}
```

```python
class Character():
    def __init__(self, payload):
        # State vars
        # Lazy record keeping
        self.char_params = ['name', 'species', 'background'] #TODO read in from a config file??
        self.char_values = None
        self._initialize(payload)

    def _validate_payload(self, payload):
        # TODO: Should this checked sentinel be something more elegant?
        checked = True
        for item in self.char_params:
            try:
                assert item in payload.keys()
            except:
                checked = False
                raise Exception(f"Required key: '{item}' is missing from payload key list.")
        return checked

    def _initialize(self, payload:dict):
    # TODO: need modification to work with front end, should we flag for console vs backend or only backend?
        if self._validate_payload(payload) :
            self.char_values = payload
        else:
            print('Check failed.')


```
