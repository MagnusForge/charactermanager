```python
# Coming from UI
test_payload = {'name':'Magnus',
       'species':'Elf',
       'background':'Crime'}

```

```python
class Character():
    def __init__(self, payload):
        # State vars
        # Lazy record keeping
        self.name = None
        self.species = None
        self.background = None
        self._initialize(payload)

    def _initialize(self, payload:dict):
        # TODO: need modification to work with front end, should we flag for console vs backend or only backend?
        self.name = payload['name']
        self.species = payload['species']
        self.background = payload['background']


```

```python
c = Character(payload=test_payload)
```
